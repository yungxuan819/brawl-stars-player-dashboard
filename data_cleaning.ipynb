```python
import pandas as pd
import numpy as np
import ast 
```

## Players table

```python
players = pd.read_csv('players.csv')
players
```

```python
# editing player_tag from O to 0
players['tag'] = players['tag'].replace('O', '0', regex=True)
players['tag'].unique()
```

```python
# cleaning 'brawlers' column

# 1. Parse the 'brawlers' column from string to list of dicts
players['brawlers'] = players['brawlers'].apply(ast.literal_eval)

# 2. Expand brawlers: 1 row per brawler, with player info repeated
players_expanded = players.explode('brawlers').reset_index(drop=True)

# 3. Normalize the 'brawlers' dictionaries into separate columns
brawlers_df = pd.json_normalize(players_expanded['brawlers'])

# 4. Drop the old 'brawlers' column and combine with the flattened brawler info
players_expanded_clean = pd.concat([players_expanded.drop(columns=['brawlers']), brawlers_df], axis=1)

players_expanded_clean.head(10)
```

```python
# Rename 'name' column
players_expanded_clean.columns = [
    'player_name' if i == 1 else 'brawler_name' if col == 'name' and i > 2 else col
    for i, col in enumerate(players_expanded_clean.columns)
]
```

```python
# Rename 'id' column
players_expanded_clean.rename(columns={'id': 'brawler_id'}, inplace=True)
```

```python
# renaming 'trophies' column
players_expanded_clean.columns = [
    'player_trophies' if i == 3 else 'brawler_trophies' if col == 'trophies' and i > 3 else col
    for i, col in enumerate(players_expanded_clean.columns)
]
players_expanded_clean.head(10)
```

```python
# renaming 'highestTrophies' column
players_expanded_clean.columns = [
    'player_highestTrophies' if i == 4 else 'brawler_highestTrophies' if col == 'highestTrophies' and i > 4 else col
    for i, col in enumerate(players_expanded_clean.columns)
]
players_expanded_clean.head(10)
```

```python
# Cleaning the 'gears' column

# 1. Parse the 'gears' column from string to list of dicts
def safe_eval(val):
    if isinstance(val, str):
        try:
            return ast.literal_eval(val)
        except (ValueError, SyntaxError):
            return []
    elif isinstance(val, list):
        return val
    else:
        return []

players_expanded_clean['gears'] = players_expanded_clean['gears'].apply(safe_eval)

# 2. Expand gears: 1 row per gear per brawler, with player info repeated
players_expanded_clean = players_expanded_clean.explode('gears').reset_index(drop=True)

# 3. Normalize the 'gears' dictionaries into separate columns
gears_df = pd.json_normalize(players_expanded_clean['gears'])

# 4. Drop the old 'gears' column and combine with the flattened gear info
players_expanded_clean = pd.concat([players_expanded_clean.drop(columns=['gears']), gears_df], axis=1)

players_expanded_clean.head(10)
```

```python
# rename 'id' 'name' 'level' columns
players_expanded_clean.rename(columns={'id': 'gear_id', 'name': 'gear_name', 'level': 'gear_level'}, inplace=True)
```

```python
# Cleaning the 'gadgets' column

# 1. Parse the 'gears' column from string to list of dicts
def safe_eval(val):
    if isinstance(val, str):
        try:
            return ast.literal_eval(val)
        except (ValueError, SyntaxError):
            return []
    elif isinstance(val, list):
        return val
    else:
        return []

players_expanded_clean['gadgets'] = players_expanded_clean['gadgets'].apply(safe_eval)

# 2. Expand gadget: 1 row per gadget per brawler, with player info repeated
players_expanded_clean = players_expanded_clean.explode('gadgets').reset_index(drop=True)

# 3. Normalize the 'gadget' dictionaries into separate columns
gadgets_df = pd.json_normalize(players_expanded_clean['gadgets'])

# 4. Drop the old 'gadget' column and combine with the flattened gadget info
players_expanded_clean = pd.concat([players_expanded_clean.drop(columns=['gadgets']), gadgets_df], axis=1)

players_expanded_clean.head(10)
```

```python
# rename 'id' 'name' columns
players_expanded_clean.rename(columns={'id': 'gadget_id', 'name': 'gadget_name'}, inplace=True)
```

```python
# Cleaning the 'starPowers' column

# 1. Parse the 'starPowers' column from string to list of dicts
def safe_eval(val):
    if isinstance(val, str):
        try:
            return ast.literal_eval(val)
        except (ValueError, SyntaxError):
            return []
    elif isinstance(val, list):
        return val
    else:
        return []

players_expanded_clean['starPowers'] = players_expanded_clean['starPowers'].apply(safe_eval)

# 2. Expand starPowers: 1 row per starPowers per brawler, with player info repeated
players_expanded_clean = players_expanded_clean.explode('starPowers').reset_index(drop=True)

# 3. Normalize the 'starPowers' dictionaries into separate columns
starPowers_df = pd.json_normalize(players_expanded_clean['starPowers'])

# 4. Drop the old 'starPowers' column and combine with the flattened starPowers info
players_expanded_clean = pd.concat([players_expanded_clean.drop(columns=['starPowers']), starPowers_df], axis=1)

players_expanded_clean.head(10)
```

```python
# rename 'id' 'name' columns
players_expanded_clean.rename(columns={'id': 'starPowers_id', 'name': 'starPowers_name'}, inplace=True)
```

```python
players_expanded_clean.head(10)
```

```python
# Remove column 'player_tag' (duplication)
players_expanded_clean.drop(columns=['player_tag'], inplace=True)
players_expanded_clean.head(10)
```

```python
# renaming 'power' and 'rank' columns
players_expanded_clean.rename(columns={'power': 'brawler_power', 'rank': 'brawler_rank'}, inplace=True)
players_expanded_clean.head(10)
```

```python
#5. Renaming 'tag' column
players_expanded_clean.rename(columns={'tag': 'player_tag'}, inplace=True)
players_expanded_clean.head(10)
```

```python
players_expanded_clean
```

```python
# Step 1: Define the core brawler ID
core_cols = ['player_tag', 'brawler_name']

# Deduplicate the entire DataFrame just to be safe
players_expanded_clean = players_expanded_clean.drop_duplicates(subset=core_cols + ['gadget_id', 'starPowers_id'])

# ----------------------
# Gadgets
# ----------------------
# Drop NaNs, then deduplicate gadgets per player-brawler combo
gadget_df = players_expanded_clean.dropna(subset=['gadget_id']).copy()
gadget_df = gadget_df.drop_duplicates(subset=core_cols + ['gadget_id'])  # dedup gadget per brawler

# Number gadgets (after deduplication)
gadget_df['gadget_num'] = gadget_df.groupby(core_cols).cumcount() + 1
gadget_df = gadget_df[gadget_df['gadget_num'] <= 2]

# Pivot to wide format
gadget_wide = gadget_df.pivot(index=core_cols, columns='gadget_num', values='gadget_id')
gadget_wide.columns = [f'gadget_id_{int(col)}' for col in gadget_wide.columns]

# ----------------------
# Star Powers
# ----------------------
sp_df = players_expanded_clean.dropna(subset=['starPowers_id']).copy()
sp_df = sp_df.drop_duplicates(subset=core_cols + ['starPowers_id'])  # dedup star_power per brawler

sp_df['sp_num'] = sp_df.groupby(core_cols).cumcount() + 1
sp_df = sp_df[sp_df['sp_num'] <= 2]

sp_wide = sp_df.pivot(index=core_cols, columns='sp_num', values='starPowers_id')
sp_wide.columns = [f'starPowers_id_{int(col)}' for col in sp_wide.columns]

# ----------------------
# Merge with brawler base
# ----------------------
brawler_base = players_expanded_clean.drop_duplicates(subset=core_cols).set_index(core_cols)
final_df = brawler_base.drop(columns=['gadget_id', 'starPowers_id', 'gadget_name', 'starPowers_name'], errors='ignore') \
                       .join(gadget_wide, how='left') \
                       .join(sp_wide, how='left') \
                       .reset_index()

# Preview
final_df
```

```python
# convert cleaned players df to csv
final_df.to_csv('cleaned_players.csv', index=False)
```

## Battle log table

```python
battlelog = pd.read_csv('battlelog.csv')
battlelog
```

```python
# editing player_tag from O to 0
battlelog['player_tag'] = battlelog['player_tag'].replace('O', '0', regex=True)
battlelog['player_tag'].unique()
```

```python
# Only apply ast.literal_eval if it's a string (to skip NaNs or already-parsed lists)
def safe_eval(x):
    if isinstance(x, str):
        try:
            return ast.literal_eval(x)
        except:
            return np.nan
    return x

# Apply safe_eval
battlelog['battle_teams'] = battlelog['battle_teams'].apply(safe_eval)

# Now define the function to extract brawler ID
def get_brawler_id(battle_teams, player_tag):
    if not isinstance(battle_teams, list):  # skip NaNs or bad data
        return np.nan
    for team in battle_teams:
        for player in team:
            if player.get('tag') == player_tag:
                return player.get('brawler', {}).get('id')
    return np.nan
```

```python
# Apply to the DataFrame, create new column 'brawler_id_used'
battlelog['brawler_id_used'] = battlelog.apply(
    lambda row: get_brawler_id(row['battle_teams'], row['player_tag']), axis=1
)
```

```python
battlelog 
```

```python
# Do the same for 'battle_players' column
# Safe evaluation again in case it's stored as a string
def safe_eval(x):
    if isinstance(x, str):
        try:
            return ast.literal_eval(x)
        except:
            return np.nan
    return x

# Apply it to battle_players
battlelog['battle_players'] = battlelog['battle_players'].apply(safe_eval)

# Function to extract brawler_id from battle_players
def get_brawler_id_2(players, player_tag):
    if not isinstance(players, list):
        return np.nan
    for player in players:
        if player.get('tag') == player_tag:
            return player.get('brawler', {}).get('id')
    return np.nan

# Apply to the DataFrame
battlelog['brawler_id_used_2'] = battlelog.apply(
    lambda row:get_brawler_id_2(row['battle_players'], row['player_tag']), axis=1
)
```

```python
battlelog 
```

```python
# Remove unneccessary columns
battlelog.drop(columns=['battle_teams', 'battle_players', 'battle_bigBrawler_tag', 'battle_bigBrawler_name', 
                        'battle_bigBrawler_brawler_id', 'battle_bigBrawler_brawler_name', 'battle_bigBrawler_brawler_power',
                        'battle_bigBrawler_brawler_trophies','battle_level_name', 'battle_level_id', 'battle_starPlayer'], 
               inplace=True)

battlelog.head(10)
```

```python
# combine brawler_id_used and brawler_id_used_2 columns
battlelog['brawler_id_combined'] = battlelog['brawler_id_used'].fillna(battlelog['brawler_id_used_2'])
battlelog 
```

```python
battlelog.drop(columns=['brawler_id_used', 'brawler_id_used_2'], inplace=True)
battlelog.head(10)
```

```python
# rename 'brawler_id_combined' to 'brawler_id_used'
battlelog.rename(columns={'brawler_id_combined': 'brawler_id_used'}, inplace=True)
```

```python
# Save cleaned battlelog DataFrame to CSV
battlelog.to_csv('cleaned_battlelog.csv', index=False)
```
