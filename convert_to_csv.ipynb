```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("brawl_data_2.db")
query = "SELECT * FROM player;"
player_df = pd.read_sql_query(query, conn)
print(player_df)
conn.close()
```

```python
# Convert player_df to CSV
player_df.to_csv("player.csv", index=False)
```
